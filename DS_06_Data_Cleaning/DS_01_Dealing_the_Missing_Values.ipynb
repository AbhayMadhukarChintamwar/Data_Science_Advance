#### 2. Replace the nan values
- Mean 
- Median 
- Mode
- fill with some value

- Continous Variables --> CLMAGE, LOSS --> Replace with either Mean or Median for continous data
- Discrete Variables --> CLMINSUR --> Mode is used for discrete data

```python
df = pd.read_csv('claimants sample.csv')
```

##### 1. fillna() using pandas

```python
mean = df.CASENUM.mean()
mean
```

```python
df['INSUR'].fillna(mean, inplace=True)
```

```python
median = df['LOSS'].median()
df['LOSS'].fillna(df['LOSS'].median(), inplace=True)
```

```python
median = df['AGE'].median()
df['AGE'].fillna(df['AGE'].median(), inplace=True)
```

```python
mode = df['INSUR'].mode()[0]
df['INSUR'].fillna(mode, inplace=True)
```

```python
df.isnull().sum()
```

##### 2. Simplemuter using sklearn

```python
df = pd.read_csv('claimants sample.csv')
df
```

```python
from sklearn.impute import SimpleImputer
```

```python
mean_imputer = SimpleImputer(strategy='mean')
df['LOSS'] = pd.DataFrame(mean_imputer.fit_transform(df[['LOSS']]))
```

```python
mean_imputer = SimpleImputer(strategy='mean')
df['AGE'] = pd.DataFrame(mean_imputer.fit_transform(df[['AGE']]))
```

```python
median_impter = SimpleImputer(strategy='median')
df['CASENUM'] = pd.DataFrame(median_impter.fit_transform(df[['CASENUM']]))
```

```python
median_impter = SimpleImputer(strategy='median')
df['INSUR'] = pd.DataFrame(median_impter.fit_transform(df[['INSUR']]))
```

```python
mode_imputer = SimpleImputer(strategy='most_frequent')
df['CASENUM'] = pd.DataFrame(mode_imputer.fit_transform(df[['CASENUM']]))
```

```python
mode_imputer = SimpleImputer(strategy='most_frequent')
df['INSUR'] = pd.DataFrame(mode_imputer.fit_transform(df[['INSUR']]))
```
