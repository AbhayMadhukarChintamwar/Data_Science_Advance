```python
import numpy as np
from scipy import stats
```

## Left Tail test

## Question

A machine is producing perfume bottles with the long term average of 150ml and the standard deviation of 2ml. Four bottles were picked and the average volume was found to be 149ml. Has the average volume min of 150ml? Test with 95% confidence level.


## Solution

 - Null Hypothesis : Mean volume  >= 150ml
 - Alternate Hypothesis : Mean volume < 150ml
 - Alpha = 1 - 0.95 = 0.05
 - **Left Tail Z test**
 - Z calculated = (Observed mean - population mean )/ standard error
 - z = (x - μ ) / (σ /√n)

```python
z_cal = (149-150)/ (2/ np.sqrt(4))
z_cal
```

```python
## Left tail ---> stats.norm.cdf(z_cal)          --> Cululative distribution finction
stats.norm.cdf(-1.0)
```

**Inferences:**
 - P > alpha
 - P High -- Null Fly
 - Do not Reject H0.

## Right Tail Test

## Question

The mean responce times for a random sample of 30 Pizza Deliveries in 32 minutes. The population standard deviation is believed to be 10 minutes. The pizza delivery services director wants to perform a hypothesis test, with 
a = 0.05 level of significance, to determine whether the service goal of 30 minutes or less being achieved.

## Solution

 - Null Hypothesis : Mean time <= 30
 - Alternate Hypothesis: Mean time > 30
 - Alpha = 0.05
 - **Right Tail Z test -- Right Tail Rejection**
 - Z calculated = (Observed mean - population mean)/ standard error
  - z = (x - μ ) / (σ /√n)


```python
z_cal = (32-30)/(10/np.sqrt(30))
z_cal
```

```python
## Right tail --> 1 - norm.cdf(z_cal)    --> Cumulative distribution function.
print(1 - stats.norm.cdf(z_cal))
```

```python
## Right tail --> 1 - norm.sf(z_cal)    --> Cumulative distribution function.

print(stats.norm.sf(z_cal))
```

**Inferences:**

- P high -- Null Fly
- Do not  reject H0
- There are not sufficient statistical evidence to infer that Pizza delivery services is not meeting the responce foal of 30 minutes.


## Two tail Test
## Question
Assume that a sample of 30 milk carton provides a sample mean of 505ml. 
- The population standard deviation is believed to be 10 ml. 
- Perform a hypothesis test, at the 0.03 level of significance, population mean 500 ml and to help determine whether the filling process should continue operating or be stopped and corrected.

## Solution

- Null Hypothesis : Mean = 500ml
- Alternate Hypothesis : Mean != 500ml
- **Two Tail Z-Test**
- Alpha = 0.03
- Z calculated = (Observed mean - population mean)/ standard error
- z = (x - μ ) / (σ /√n)


```python
z_cal = (505-500)/(10/ np.sqrt(30))

print(z_cal)
```

```python
# Two tail --> (norm.sf(z_cal))*2

pvalue = 2 * stats.norm.sf(z_cal)
print(pvalue)
```

**Inferences:**

- P low -- Null go
- Reject H0
- There is no sufficient statistical evidence to infer that the null hypothesis is true. (i.e. the mean filling quantity is not 500ml)
