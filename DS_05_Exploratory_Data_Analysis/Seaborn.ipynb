## Histogram / Distrubution plot
 Dist plot helps us to check the distribution of a feature

```python
sns.displot(df.total_bill)
plt.show()
```

```python
sns.displot(df.total_bill, bins = 10, kde= True)
plt.title('Histogram of total_bill')
plt.show()
```

## Box Plot
A box and whisker plot (something called a boxplot) is a graph that presents information from a five-number summary.

```python
sns.boxplot(x = df.total_bill)
plt.show()
```

## Join Plot

A join plot allows to study the relationship between 2 numeric variables. The central chart display their correlation. It is usually a scatterplot, a hexbin plot, a 2D histogram or a 2D density plot

```python
sns.jointplot(y = 'tip', x = 'total_bill', data = df)
plt.show()
```

## Violin Plot
Violin plot helps us to see  both the distribution of data in terms of Kernel density estimation and the box plot

```python
sns.violinplot(x = 'total_bill', y = 'day', data= df)
plt.show()
```

```python
sns.violinplot(x='total_bill' , y = 'day', data = df, palette = 'rainbow')
plt.show()
```

## Pair Plot - Multiple continous variables

 A 'pairs plot' is also know as a scatterplot, in which one variable in the same data row is matched with another variable's value, like this: Pairs plots are just elaboration on this, showing all variables paired with all the other variables.

```python
sns.pairplot(df)
plt.show()
```

```python
continous = ['total_bill', 'tip']
sns.pairplot(df, vars = continous)
plt.show()
```

```python
continous = ['total_bill', 'tip']
sns.pairplot(df, vars = continous, hue = 'sex')
plt.show()
```

## Heat Map
A heatmap uses colored cells to represent relation between variables

### Heatmap (for correlation)

- A correlation heatmap uses cells to show a 2D correlation matrix (table) between two numeric dimensions.
- It is very important in Feature Selection.

**Correlation (r)** = COV(x,y)/(σxσy)

```python
corr_matrix = df[['total_bill', 'tip']].corr()
corr_matrix
```

```python
sns.heatmap(corr_matrix, annot = True)
plt.show()
```
