## Encoding

**Converting Discrete Categorical Variable to  Discrete Numberiacal Variable**
**There are 2 types of categorical variables**

- 1.Nominal (Ex: Item type)
- 2.Ordinal (Ex: Outlet_Size (Small, Medium, High))

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('homeprices.csv')
df
```

## To convert Nominal Categorical into Numeric

1. **get_dummies() using Pandas**

```python
df = pd.get_dummies(df, drop_first=True)
df
```

### 2. OneHotEncoding Using sklearn

```python
df = pd.read_csv('homeprices.csv')
print(df.head())

## OneHotcoding : We use the OneHotEncoder from sklearn library

from sklearn.preprocessing import OneHotEncoder

## call the function
enc = OneHotEncoder(drop= 'first')

## fit_transform

enc_df = pd.DataFrame(enc.fit_transform(df[['town']]).toarray())
enc_df
```

```python
# merge with main df

df_ohe = pd.concat([df, enc_df],axis='columns')

# drop the original variable

df_ohe.drop('town', axis = 'columns', inplace =True)


df_ohe
```

## To convert Ordinal Categorical into Numberic 


#### 1. LabelEncoder

- convert to numeric as per alphabetical order

```python
df = pd.read_csv('homeprices.csv')


from sklearn.preprocessing import LabelEncoder

label = LabelEncoder()

df['town'] = label.fit_transform(df.town)
df
```

#### 2. OrdinalEncoder
- convert to numeric as per given order (ascending order) in the function

```python
df = pd.read_csv('homeprices.csv')

from sklearn.preprocessing import OrdinalEncoder


oe = OrdinalEncoder(categories=[['Bangalore','Hyderabad','Chennai']])

df.town = oe.fit_transform(df[['town']])
df
```

#### 3. map() using pandas

- convert to numeric as per your choice

```python
df = pd.read_csv('homeprices.csv')

df['town'] = df['town'].map({
    'Bangalore':0,
    'Hyderabad':1,
    'Chennai':2
})

df
```
