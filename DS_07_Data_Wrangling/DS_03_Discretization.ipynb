## Discretization

Discretization is the proocess of transformation continuous variables into discrete variables by creating a set of contiguous intervals that span the range of the variable's values. Discretization is also called **binning**, where bin is an alternative name for interval.

#### Discretization helps handle outliers and may improve value spread in skewed variables.

Discretization helps handle outliers by placing these values into the lower intervals, together with the remaining inlier values of the distribution. Thus, these outlier observations no longer differ from the rest of the values at the tails of the distribution, as they are now all together in the same interval / bucket. In addition, by creating appropriate bins or intervals, discretization can help spread the values of a skewed variable across a set of bins with equal number of observations.

#### Creating Bins

```python
intervals = [0,12,19,30,60, 90]

categories = ['child','teenager', 'young_adult', 'middle_aged','senior_citizen']
```

```python
stroke_data['age_category'] = pd.cut(x = stroke_data['age'],
                                                     bins = intervals,
                                                     labels = categories)
```
