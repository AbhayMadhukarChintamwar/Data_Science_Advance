## Feature Preprocessing : Tranformation

```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('titanic.csv')
data.head()
```

```python
data = pd.read_csv('titanic.csv', usecols=['Fare'])
data.head()
```

```python
data.Fare.skew()
```

## Root Transformation

```python
data['sqr_Fare'] = data.Fare **(1/4)
data.sqr_Fare.skew()
```

## Logarithmi Transformation

```python
data['Log_Fare'] = np.log(data.Fare + 0.01)
data.Log_Fare.skew()
```

## Reciprocal Transformation

```python
data['Rec_Fare'] = 1/ (data.Fare + 0.01) 
data.Rec_Fare.skew()
```

## BoxCox

The Box-Cox transformation is defined as:

(X**λ -1)/λ

where X is the response variable and λ is the tranformation parameter. λ variable from the transformation, all values of λ are consider and the optimal  value for a given variable selected.


```python
from scipy import stats
```

```python
# transform training data & save Lambda value
data['Fare_boxcox'], param = stats.boxcox(data.Fare + 0.01)
```

```python
print('λ = ', param)
```

```python
data.Fare_boxcox.skew()
```
