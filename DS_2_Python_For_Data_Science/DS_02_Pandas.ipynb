## Pandas Series

1. Create Series using List

```python
import numpy as np
import pandas as pd

l = ['banana', 42]

s = pd.Series(l)
print(s)

```

```python
s = pd.Series(l, index=['a','b'])
print(s)
```

**2. Create Series using Array**

```python
a = np.array(l)

s = pd.Series(a, index= ['a', 'b'])
print(s)
```

**3. Create Series using Dictionary**

```python
d = { 'a': 1, 'b':2, 'c':3}
ser = pd.Series(d)
print(ser)
```

```python
d = { 'a': 1, 'b':2, 'c':3}
ser = pd.Series(d, index= ['a','c'])
print(ser)
```

```python
d = {'col1':[1,2], 'col2':[3,4]}
df = pd.Series(d)
print(df)
```

## Data Frame --> pandas.DataFrame

**1. Create DataFrame using Dictionary**

```python
d = {'col1':[1,2], 'col2':[3,4]}
df = pd.DataFrame(d)
print(df)
```

**2. Create DataFrame using Nested List or 2D array**

```python
temperature_df = pd.DataFrame({
    'city':['Mumbai','Delhi','Hyderabad','Banglore'],
    'temperature':[32, 45, 34, 36]
})

print(temperature_df)
```

**Attrubutes**

```python
# Displaying the dimensions of the data frame using shape.
temperature_df.shape
```

```python
# Displaying the size of the data frame using size.
temperature_df.size
```

```python
# Displaying the name of the coloumns
temperature_df.columns
```

```python
# Displaying the name of the columns using keys() function.
temperature_df.keys()
```

```python
temperature_df.dtypes
```

**Add a column**

```python
# adding column named 'AQI' tp the dataframe.
temperature_df['AQI']= [120, 150, 100, 80]
print(temperature_df)
```

**Drop a row**

- Drop (Need to specify that axis = 0 or  'index', axis = 1 or 'columns')

```python
# Deleting row from the data frame.

df = temperature_df.drop(index=[0])
print(df)
```

```python
# Deleting multiple rows from the data frame
df= temperature_df.drop(index =[0,1])
df
```

**Drop a column**

- Drop (Need to specify that axis = 0 or 'index', axis = 1 or 'columns')

```python
# Deleting multiple columns from the data frame using drop() function.
df = temperature_df.drop(columns=['AQI', 'city'])
df
```

```python
# Deleting multiple columns from the data frame using drop function
df = temperature_df.drop('AQI', axis = 1)
df
```

```python
temperature_df = temperature_df.drop('AQI', axis = 1)
temperature_df
```

```python
temperature_df
```

```python
temperature_df.drop('city', inplace=True, axis =1)
temperature_df
```

```python
temperature_df = pd.DataFrame({
    'city':['Mumbai','Delhi','Hyderabad','Banglore'],
    'temperature':[32, 45, 34, 36]
})

print(temperature_df)
```

```python
humidity_df = pd.DataFrame({
    'city':['Mumbai','Hyderabad','Banglore'],
    'humidity':[68, 65, 74]
})

print(humidity_df)
```

### Combining Data frames

option 1: **Append**

```python
# Adding rows to the data frame by adding other data frame
df = temperature_df.append(humidity_df)
df
```

```python
# Adding rows to the data frame by adding other data frame
df = pd.concat([temperature_df,humidity_df])
df
```

```python
# Adding rows to the data frame by adding other data frame
df = pd.concat([temperature_df,humidity_df], axis = 1)
df
```

```python
# Inner join
df = pd.merge(temperature_df, humidity_df, on= 'city', how='inner')
df
```

```python
# Outer join
df = pd.merge(temperature_df, humidity_df, on='city', how='outer')
df
```
