## NumPy

NumPy (**Num**eriacal **Py**thon) is an open source Python library that aids in mathemaical and scientific programming.

NumPy is the fundamental package for scientific computing with Python, It contains:

* a powerful N-dimensional array object
* extension package to Python for multi-dimensional arrays
* designed for scientific computation -- linear algebra and statistical oparations capabilities
* NumPy is increadible library to perform mathematical and statistical operations due to it's fast and memory effcient.
* This allows NumPy to seamlessly and speedily integrate with a wide variety of databases.


```python
##Import NumPy
import numpy as np
```

## Creating NumPy Arrays From a List

**1 - D**

```python
a = np.array([0, 1, 2, 3])
print(a)
print(a.ndim)    # print diamesions
print(a.shape)   #print shape
```

**2 - D**

```python
b = np.array([[0, 1, 2], [3, 4, 5]])
print(b)
print(b.ndim)
print(b.shape)
```

```python
# We can Convert a matrix which is also a list to array

my_matrix = [[1,2,3],[4,5,6],[7,8,9]]
print(my_matrix)
b = np.array(my_matrix)
print(b)
print(b.ndim)
print(b.shape)
```

#### Functions for creating arrays

**arange**
- arange is an array-valued version of the built-in Python range function
- **Syntax:**  np.arange(start, end , step)

```python
np.arange(10, 100, 10)   # Start, end (exclusive), step
```

**linspace**
- Return evenly spaced numbers over a specified interval.
- **Syntax:** np.linspace(start, end, num_of_samples)

```python
np.linspace(0, 10, 5)
```

**zeros**
- Generate array of zeros
- **Syntax:** np.zeros(shape)

```python
np.zeros((3))
```

```python
np.zeros((3,3)) # zero matrix or null matrix
```

**random.randint**
- return random integers from start(inclusive) to end (exclusive)
- **Syntax:** np.random.rad=ndint(start, end, num_of_samples) 

```python
np.random.randint(1, 41, 5)
```

### Operations

**Flattening :** Converting nd array to 1D array

```python
a = np.array([[1,2,3],[4,5,6]])
print(a)

a.ravel()
```

**Reshape array**

```python
arr = np.arange(20)
print(arr)
print(arr.shape)
```

```python
# Notice the two sets of brackets
a1 = arr.reshape(5,4)
a1
```

```python
a = np.array([[1,2,3], [4, 5,6]])
print(a)
```

```python
b = a.reshape(3,2)
print(b)
```

**Transpose array**

```python
a = np.array([[1,2,3],[4,5,6]])
print(a)
```

```python
c = a.T
print(c)
```

**Copy array**

```python
a = np.array([1, 2, 3])
b = a    # shallow copy
b[0] =100
print(a)
print(b)
```

```python
a = np.array([1,2,3])
b = a.copy()               #carbon copy or deep copy
b[0] = 100
print(a)
print(b)
```

**Sort**

```python
np.sort([4,5,1,6,9,8,2,3,7,0])
```

```python
# Sorting along an axis: axis = 1---> row and axis = 0 ---> column

a= np.array([[5,4,6],[2,8,2]])
b = np.sort(a)
b
```

**Add/Remove operations on a numpy array**

```python
a = np.array([4,7,8])

print('Numpy array before addition :',a)

a = np.append(a, [2,3,4])
print('Numpy array after addition :',a)

a = np.delete(a, 2)   # deleting index[2] element  i.e. 8
print('Numpy array agter deletion of one element: ',a)   
```
