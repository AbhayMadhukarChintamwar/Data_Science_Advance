## Data Extraction

this can be done by 3 ways

- 1. By using conditions
- 2. by using iloc() & loc()
- 3. combination of condition & loc()


**using 1 condition**

```python
df['Gender']=='Male'
```

```python
df[df['Gender']=='Male']
```

```python
male_data = df[df['Gender']=='Male']
male_data
```

```python
df[df['Age']>=50]
```

**Multiple Conditions**

```python
filter1 = df[(df['Age']==65)|(df['Gender']=='Female')]
filter1
```

```python
filter2 = df[(df['LiverPatient']==1)&(df.Age>=70)]
filter2
```

```python
filter3 = df[(df['Gender']=='Female')|(df.Age>=35)| (df['DB']<=6)]
filter3
```

```python
# Using = condition for selected columns

df1= df.loc[df['DB']>= 3.7, ['Age','DB','LiverPatient']]
df1['LiverPatient'].value_counts()
```

```python
# Using  < condition for select columns

df.loc[df['TB']<0.5, 'Gender':'LiverPatient']
```

```python
# Using  condition for selected columns

df.loc[df.Age>80, 'Age':'LiverPatient']
```

```python
# Using > and < conditions together (using &) for selected columns

df.loc[(df.Sgpt>400)&(df.Sgpt<=420), 'TB':'Alkphos']
```
