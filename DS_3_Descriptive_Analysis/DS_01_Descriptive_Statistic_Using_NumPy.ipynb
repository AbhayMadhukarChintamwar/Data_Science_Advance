# Mean Absolute Deviation 

```python
x =np.array(np.random.randint(1, 20, 5))
x
```

```python
deviation = x - x.mean()
deviation
```

```python
abs_deviation = abs(deviation)
abs_deviation
```

## Mean Square Deviation

```python
x = np.array(np.random.randint(1, 20, 5))
print('x :', x)
deviation = x-x.mean()
print('Deviation: ', deviation)
sqr_deviation = deviation * deviation
print('Square Deviation: ', sqr_deviation)
print('Mean Square Deviatioin: ', np.mean(sqr_deviation))
```

## Sample Variance & Sample Standard

## Deviation

```python
s = np.array(np.random.randint(1000, 2500, 4))
s
```

```python
sample_variance = np.var(s) * len(s)/ len((s-1))
sample_variance
```

```python
sample_std_dev = np.sqrt(sample_variance)
sample_std_dev
```
