## Matplotlib

Matplotlib is a plotting library for the Python programming language and its numerical mathematics extensionn NumPy.

Some of the major Pros of Matplotlib are:

- Generally easy to get started for simple plots.
- Support for custom labels and texts
- Great control of every element in a figure
- High-quality output in many formats
- Very customizable in general

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
d = {'Length': np.random.randn(500)}

df = pd.DataFrame(d)
df
```

## Histogram

**In histogram, x axis contains a variable and y axis will be a frequency of that variable**

```python
df.Length.hist()    #using pandas
```

```python
plt.hist(df.Length)      # 10 bins by default
plt.xlabel('Length')    # set the label for the x-axis
plt.ylabel('Count/Frequency')  # set the label for the y-axis
plt.title('Histogram of Length')
plt.show()
```

```python
plt.hist(df.Length, bins=15)
plt.show()
```

```python
# Horizontal orientation

plt.hist(df.Length, orientation='horizontal')
plt.xlabel('count')
plt.ylabel('length')
plt.title('Histogram of length')
plt.show()
```

```python
days = [1,2,3,4,5,6,7]
max_t = np.random.randint(45, 55, 7)
print(max_t)
min_t = np.random.randint(35, 45, 7)
print(min_t)
avg=[]
for i in range(0, 7):
     avg_t = (max_t[i]+ min_t[i])/2
     avg.append( int(avg_t))
print(avg)


plt.plot(days, max_t, label = 'max')
plt.plot(days, min_t, label = 'min')
plt.plot(days, avg, label = 'average')

# set axes labels and chart title

plt.xlabel('Day')
plt.ylabel('Temperature')
plt.title('Weather')

# set Legend
plt.legend()
plt.show()
```

## Subplots

```python
x = []
y =np.random.randint(45, 50,7)
for i in range(1, 8):
    x.append(i)

print(x)
print(y)

plt.subplot(3,2,1)
plt.plot(x, y, color='green', marker = '*', linestyle = 'dashed')
plt.title('line plot')

plt.subplot(3,2,2)
plt.scatter(x, y)
plt.title('scatter plot')

plt.subplot(3,2,3)
plt.plot(x, y, linewidth= 3)

plt.subplot(3,2,4)
plt.plot(x, y, color = 'green', marker = 'o', linestyle = 'dashed')

plt.show()
```

## Pie Chart

```python
list = np.random.randint(0, 40, 6)
lbs = ['FaceCream', 'FaceWash','ToothPaste','Bathing soap','Shampoo','Moisturizer']
plt.pie(list, labels = lbs, autopct = '%1.1f%%', radius = 1.5)
plt.show()
```

```python
# counterclock and angle properties
plt.pie(list, labels = lbs, autopct='%1.1f%%', explode=[0, 0.1, 0, 0, 0, 0])
plt.axis('equal')
plt.show()
```

```python
# Data to plot
la = 'Python', 'C++', 'Ruby', 'Java'
sizes = [50, 10, 5, 35]
co = ['gold','yellowgreen','lightcoral', 'lightskyblue']
explode = (0.1, 0, 0, 0)

#plot
plt.pie(sizes, explode= explode, labels = la, colors=co, autopct='%1.1f%%')

# saving the plot

plt.savefig('Plot.jpg')
plt.show()
```

```python
# sinmple bar chart showing revenues of major US tech companies
company = ['Google', 'Amazon','Microsoft','Facebook']
revenue = [90, 136, 89, 27]

plt.bar(company, revenue)
plt.ylabel('Revenue($B)')
plt.title('US Technology Stocks')
plt.show()
```

```python
#Hosizontal bar chart using barh function

plt.barh(company, revenue)
plt.ylabel('Revenue($B)')
plt.title('US Technology Stocks')
plt.show()
```

## Stacked Bar plot

```python
company = ['Google', 'Amazon','Microsoft','Facebook']
revenue = [90, 136, 89, 27]
profit = [40, 20, 34, 12]
plt.bar(company, revenue, label = 'Revenue')
plt.bar(company, profit, label = 'Profit')

plt.title('US Technology Stocks')
plt.legend()
plt.show()
```

- Revenue wise ---> Amazon is high
- Profit wise --> Google is high

```python
company = ['Google', 'Amazon','Microsoft','Facebook']
revenue = [90, 136, 89, 27]
profit = [40, 20, 34, 12]

xpos = np.arange(len(company))
plt.bar(xpos-0.2, revenue, width=0.4, label= 'Revenue')
plt.bar(xpos+0.2, profit, width=0.4, label = 'Profit')

plt.xticks(xpos, company)
plt.ylabel('Revenue($B)')
plt.title('US Technology Stocks')
plt.legend()
plt.show()
```

```python
# Program for drawing multiple bar charts on one image.

ecommerce = ['Myntra','Snapdeal','Alibaba','Amazon','Flipkart']
Q1_Profit = np.random.randint(20, 150, 5)
Q2_Profit = np.random.randint(20, 150, 5)
Q3_Profit = np.random.randint(20, 150, 5)
Q4_Profit = np.random.randint(20, 150, 5)

print(Q1_Profit)
print(Q2_Profit)
print(Q3_Profit)
print(Q4_Profit)
```

```python
# Creating different bar charts on one image using subplot() function

plt.figure(1, figsize= (10, 10))

# Creating bar chart in first cell of figure having 2 rows, 3 columns.
plt.subplot(2,2,1)
plt.bar(ecommerce, Q1_Profit)
plt.title('Quater1 Profit')

# Creating bar chart in second cell

plt.subplot(2,2,2)
plt.bar(ecommerce, Q2_Profit)
plt.title("Quater2 Profit")

# Creating a bar chart in third cell.
plt.subplot(2,2,3)
plt.bar(ecommerce, Q3_Profit)
plt.title("Quater3 Profit")

# Creating a bar chart in fourth cell.
plt.subplot(2,2,4)
plt.bar(ecommerce, Q4_Profit)
plt.title("Quater4 Profit")


# Adding a Main title for the figure
plt.suptitle('Profit on Quaterly Basis')


# Displaying the chart
plt.show()
```
