```python
import numpy as np
import pandas as pd
import tensorflow as tf
import keras
import warnings
warnings.simplefilter('ignore')
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

## **preprocessing the training set**

```python
gen = ImageDataGenerator( rescale = 1/255, shear_range = 0.2, zoom_range = 0.2)
```

```python
training_set = gen.flow_from_directory('dataset\\training_set', target_size=(64, 64), class_mode = 'binary')
```

```python
training_set.class_indices
```

```python
test = ImageDataGenerator(rescale=1/255)
```

```python
test_set = test.flow_from_directory('dataset\\test_set', target_size=(64,64), class_mode='binary')
```

## **modeling built convolution neural network**

```python
from keras.models import Sequential
```

```python
cnn = Sequential()
```

```python
from keras.layers import Conv2D
```

```python
cnn.add(Conv2D(filters=32, kernel_size = 3,activation = 'relu', input_shape = [64,64,3]))
```

```python
from keras.layers import MaxPooling2D
```

```python
cnn.add(MaxPooling2D(pool_size=2, strides = 2))
```

## **adding convolution layer**

```python
cnn.add(Conv2D(filters=32, kernel_size = 3,activation = 'relu'))
cnn.add(MaxPooling2D(pool_size = 2 , strides = 2))
```

## **flattening**

```python
from keras.layers import Flatten
```

```python
cnn.add(Flatten())
```

## **full connection**

```python
from keras.layers  import Dense
```

```python
cnn.add(Dense(units = 128,activation = 'relu'))
cnn.add(Dense(units=1 , activation = 'sigmoid'))
```

## **compailing**

```python
cnn.compile(optimizer = 'adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
cnn.fit(x=training_set,validation_data = test_set, epochs=25)
```

## **Evaluation**
