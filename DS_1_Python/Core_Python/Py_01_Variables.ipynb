### Varible
 - A variable is a location in memory used to store data(value)
 - In python, variables are Dynamic (We don't even have to declare the type of the variable).


 ##### Rules in Naming a Variable
 - Rule 1: A Variable name must start with a letter
 - Rule 2: A Variable name cannot start with a number
 - Rule 3: A Variable name can only contain alphanumeric charecters
 - Rule 4: A Variable accepts only underscore "_" symbol only (No other symbol is accepted)
 - Rule 5: Variable names are Case Sensitive (age, Age and AGE are three different variables)
 - Rule 6: It can't accept reserve/keywords

## Variable Assignment

```python
#  this is first variable assignment
a = 9
print(type(a),a)

A = 2.0
print(type(A),A)

z = 'amc'
print(type(z),z)

ch = 'c'
print(type(ch), ch)
```

### Assign Multiple Values to Multiple Variables at one time

```python
# We can assign multiple variables in a single line
a,b,c  = 10,20,30
print(a,b,c)
```

```python
# We can assign the same value to multiple variables at once
a = b = c = 'AI ML Engineer'
print(a, b,c)
```
