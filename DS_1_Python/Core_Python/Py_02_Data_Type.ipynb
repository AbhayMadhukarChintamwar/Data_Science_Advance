### Type Casting: Convert from one data type to another data type

```python
b = 5.9
print(b)
print(type(b))

print(int(b))

print(int(10.9999))
print(float(5.4)) 
```

## Boolean - True/ False

```python
b1 = True
print(b1)
print(type(b1))
print(int(True))
print(int(False))
print(bool(-1.0))
print(bool(0))
```

## Complex

Complex literals can be created by using the notation x +yj , where x is the real component and y is the imaginary component.

```python
x = 10 + 20.5j

print(type(x))
# complex

# to get real value

print(x.real)
# 10

#to get imaginary value

print(x.imag)
# 20.5


```

```python
int(10+5j)   # TypeError : Can't convert complex to int


```

```python
complex(10.4)
(10.4+0j)
```

## String

- A string in python consist of a series or sequence of characters - letters, number, and special characters.
- We can use single quotes or double quotes to represent strings.

```python
a = 'siva'
print(a)
# siva

type(a)
str


myString1 = 'Hello'
print(myString1)
# Hello

myString2 = "Hello"
print(myString2)
# Hello

myString3 = '''Hello'''
print(myString3)
# Hello
```

- Multiline strings can be denoted using triple quotes, ''' or """


```python
name = '''siva
rama
krishna'''
print(name)

# siva
# rama
# krishna

var1 = 1234567
type(var1)
# int

var = '1234567'
type(var)
# str

print(float(var))
# 1234567.0
```

```python

a = 'si@1234'
print(int(a))               # ValueError : invalid literal for int() with base 10 : 'si@1234'
```
