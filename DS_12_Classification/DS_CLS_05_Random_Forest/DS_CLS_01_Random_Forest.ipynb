```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('penguins_size.csv')
data.head()
```

```python
data.info()
```

```python
data.species.value_counts()
```

```python
data.species.unique()
```

```python
data.species = data.species.replace({
    'Adelie':0,
    'Chinstrap':1,
    'Gentoo':2
}).astype('int')
```

```python
data.island.unique()
```

```python
data.island = data.island.replace({
    'Torgersen':2, 'Biscoe':0, 'Dream':1
}).astype('int')
```

```python
data.sex.unique()
```

```python
data = data[data.sex != '.']
```

```python
data.sex.unique()
```

```python
data.sex = data.sex.fillna(data.sex.mode()[0])
```

```python
data.sex = data.sex.replace({
    'MALE':1,
    'FEMALE':0
}).astype('int')
```

```python
data.isnull().sum()
```

```python
data.culmen_length_mm = data.culmen_length_mm.fillna(data.culmen_length_mm.mean())
data.culmen_depth_mm  = data.culmen_depth_mm.fillna(data.culmen_depth_mm.mean())   
data.flipper_length_mm = data.flipper_length_mm.fillna(data.flipper_length_mm.mean())
```

```python
data.body_mass_g = data.body_mass_g.fillna(data.body_mass_g.mean())
```

```python
pd.crosstab(data['species'],data['island'],margins = True)
```

```python
pd.crosstab(data['species'],data['sex'],margins = True)
```

```python
data.describe()
```

## **encoding**

```python
pd.get_dummies(data.drop('species', axis = 1), drop_first=True)
```

```python
X = data.drop('species', axis = 1)
y = data['species']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=101, test_size=0.3)

from sklearn.ensemble import RandomForestClassifier

dtc = RandomForestClassifier(random_state=42)
dtc.fit(X_train, y_train)

train_pred = dtc.predict(X_train)
test_pred = dtc.predict(X_test)

# accuracy

from sklearn.metrics import accuracy_score
print(accuracy_score(y_train,train_pred))
print(accuracy_score(y_test,test_pred))


# cross validation score

from sklearn.model_selection import cross_val_score

score = cross_val_score(dtc, X, y, cv= 5)
print(score)

print(score.mean())

from sklearn.metrics import confusion_matrix
print(confusion_matrix(y_test, test_pred))
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test,test_pred))
```

## **Hyperparameter tuning**

```python
from sklearn.model_selection import GridSearchCV

estimator = RandomForestClassifier(random_state=101 , bootstrap=False)
param_grid = {'n_estimators':list(range(1,10))}

Hp_rfc = GridSearchCV(estimator, param_grid, cv= 5, scoring='accuracy')

Hp_rfc.fit(X_train, y_train)

Hp_rfc.best_params_
```

## **Rebuild the final model**

```python
rtc = RandomForestClassifier(n_estimators = 5, bootstrap=False)
rtc.fit(X_train, y_train)

train_pred = dtc.predict(X_train)
test_pred = dtc.predict(X_test)

print('Train accuracy:' , accuracy_score(y_train,train_pred))
print('Test accuracy:' , accuracy_score(y_test,test_pred))

score= cross_val_score(dtc,X,y,cv=5)
print(score)
score.mean()
```
